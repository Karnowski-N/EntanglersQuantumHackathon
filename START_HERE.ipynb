## Our main extension

[Reduced-qubit hybrid hourly](solution/reduced_qubit_hybrid_hourly.ipynb) contains the core method, qubit-reduction checks, 5/10-plant comparisons, dispatch and seed/shot study. Selected hours are **13, 14, 15, 16, 17**: the endpoint 18:00 is excluded. Smaller circuits and improvements over random sampling do not demonstrate an advantage over MILP.

## Supporting files

`src/` contains the Python implementation. `tests/` contains checks. `data/` contains the input fleets and demand. `results/` contains recorded benchmark and hardware evidence. `docs/` contains assignment and background references.

Run tests from the project root with `python -m unittest discover -s tests -t .` in the environment specified by `requirements.txt`. Run command-line tools as `python src/<script>.py --help`. Existing recorded source hashes describe the code at experiment time; moving the code does not rewrite historical evidence.
